```python
from sklearn.datasets import fetch_openml
import numpy as np
from cech_umap.cumap import CechUMAP
import matplotlib.pyplot as plt
```

```python
D = fetch_openml("mnist_784", as_frame=False)
X = (D.data.astype(np.float32) / 255.0)
y_raw = D.target
y = y_raw.astype(np.int64) if y_raw.dtype.kind in "i" else np.array(y_raw, dtype=str).astype(int)
```

```python
cumap = CechUMAP(n_neighbors = 15)
emb = cumap.fit_transform(X)
```

```python
fig,ax = plt.subplots(figsize=[15,15])
plt.scatter(emb[:,0],emb[:,1],c = y, cmap = 'tab10')
plt.show()
```
